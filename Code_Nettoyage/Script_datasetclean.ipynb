```python

# charger le fichier Excel et lire la première feuille
df = pd.read_excel("Data/data_brute/Chiffres sur la production, les importations et exportations de poissons et de viande.xlsx")

# Afficher les premières lignes
df.head()

```

```python
# afficher les informations sur le DataFrame
df.info()

# afficher les statistiques descriptives
df.describe()

# verifier les colonnes du DataFrame
df.columns
```

```python
# verifier les valeurs manquantes
df.isnull().sum()
```

```python
df[df['Année'].isnull()].head()
df[df['Type'].isnull()].head()
df[df['Valeur'].isnull()].head()
```

```python
# Remplacer le type float de la date par un type int64
df['Année'] = df['Année'].astype('Int64')  # supporte les valeurs manquantes
```

```python
# Remplacer le type de valeur de objet en float
df['Valeur'] = pd.to_numeric(df['Valeur'], errors='coerce')

#verifaction du type de la colonne
print(df['Valeur'].dtypes)  # Doit afficher: float64
```

```python
# Traitement des valeurs manquantes

#Remplacer les valeurs manquantes pour Année
mode_annee = df['Année'].mode()[0]  # Trouve l'année la plus fréquente
df['Année'].fillna(mode_annee, inplace=True)

# Remplacer les valeurs manquantes pour Type
df['Type'].fillna(df['Type'].mode()[0], inplace=True)

#Remplacer par la médiane (robuste aux valeurs extrêmes)

mediane_valeur = df['Valeur'].median()
df['Valeur'].fillna(mediane_valeur, inplace=True)

#Vérification après remplacement
df.isnull().sum()
```

```python
# Afficher les premières lignes
df.head()
```

```python
# Définir le chemin de sauvegarde
chemin_fichier = r"Data/datacleaned/Chiffres_production_importations_exportations de poissons et de viande.json"

# Sauvegarder le DataFrame en format JSON
df.to_json(chemin_fichier, orient='records', lines=True, force_ascii=False)

# Confirmation
print(f"Le fichier a été sauvegardé sous : {chemin_fichier}")
```

```python
# importation des bibliothèques nécessaires pour l'analyse de données
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
# Chargement du 1er dataset
df1 = pd.read_excel("Data/data_brute/Indicateurs provenant de la Banque Mondiale.xlsx")

# Chargement du 2e dataset
df2 = pd.read_excel("Data/data_brute/Marché de l'internet et de la téléphonie mobile de 2010 à 2016.xlsx")

# Affichage rapide des premières lignes pour vérification
print("Premier dataset : Indicateurs provenant de la Banque Mondiale ")
display(df1.head())

print("Deuxième dataset: Marché de l'internet et de la téléphonie mobile de 2010 à 2016")
display(df2.head())
```

```python
# Analyse exploratoire (EDA)
```

```python
# Comparer les structures

print("Structure du dataset 1 :")
display(df1.info())

print("Structure du dataset 2 :")
display(df2.info())
```

```python
#  Analyse descriptive

print("Statistiques descriptives du dataset 1 :")
display(df1.describe())

print("Statistiques descriptives du dataset 2 :")
display(df2.describe())
```

```python
# verifier les colonnes du DataFrame

print("Colonnes du dataset 1 :")
display(df1.columns)

print("Colonnes du dataset 2 :")
display(df2.columns)
```

```python
# Nettoyage et uniformisation

#les colonnes en minuscules et sans espace 
df1.columns = ['annee', 'indicateur', 'valeur']
df2.columns = ['annee', 'categorie', 'sous_categorie', 'entreprise', 'statut', 'valeur']
```

```python
# Nombre de valeurs manquantes par colonne

print("Valeurs manquantes dans df1 :")
print(df1.isnull().sum())

print("\nValeurs manquantes dans df2 :")
print(df2.isnull().sum())

# -----------------------------------------------------------------------------------------------
# Pourcentage de valeurs manquantes par colonne

# print("Pourcentage de valeurs manquantes dans df1 :")
# print((df1.isnull().mean() * 100).round(2))

# print("\nPourcentage de valeurs manquantes dans df2 :")
# print((df2.isnull().mean() * 100).round(2))

# -----------------------------------------------------------------------------------------------

# Afficher les lignes contenant des valeurs manquantes

print("Lignes avec valeurs manquantes dans df1 :")
display(df1[df1.isnull().any(axis=1)])

print("Lignes avec valeurs manquantes dans df2 :")
display(df2[df2.isnull().any(axis=1)])

```

```python
# Remplissage des valeurs manquantes dans df1
df1['valeur'].fillna(df1['valeur'].median(), inplace=True)

# Remplissage des valeurs manquantes dans df2

df2['sous_categorie'].fillna(df2['sous_categorie'].mode()[0], inplace=True)
df2['valeur'].fillna(df2['valeur'].median(), inplace=True)

```

```python
# verification des valeurs manquantes après le remplissage
print("Valeurs manquantes restantes dans df1 :\n", df1.isna().sum())
print("Valeurs manquantes restantes dans df2 :\n", df2.isna().sum())
```

```python
# Nettoyage final

# Vérification des doublons

print("Nombre de doublons dans df1 :", df1.duplicated().sum())
print("Nombre de doublons dans df2 :", df2.duplicated().sum())
 
```

```python
# Suppression des doublons dans df1
df1.drop_duplicates(inplace=True)
print("Nombre de doublons dans df1 :", df1.duplicated().sum())
```

```python
# Affichage rapide des premières lignes pour vérification
print("Premier dataset : Indicateurs provenant de la Banque Mondiale ")
display(df1.head())

# affichage des 10 dernières lignes du premier dataset
print("10 Dernières lignes du premier dataset :")
display(df1.tail(10))

#----------------------------------------------------------------------------------

print("Deuxième dataset: Marché de l'internet et de la téléphonie mobile de 2010 à 2016")
display(df2.head())

# affichage des 10 dernières lignes du deuxième dataset
print("10 Dernières lignes du deuxième dataset :")
display(df2.tail(10))
```

```python
# VISUALISATIONS après le nettoyage et transformation des données

#importation des bibliothèques nécessaires pour la visualisation
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Pour la dataframe df1

# Évolution de la valeur des indicateurs dans le temps df1
plt.figure(figsize=(12, 6))
sns.lineplot(data=df1, x='annee', y='valeur', hue='indicateur') 
plt.title('Évolution de la valeur des indicateurs dans le temps')
plt.xlabel('Année')
plt.ylabel('Valeur')
plt.legend(title='Indicateur', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid()
plt.show()


# visualiser l'évolution d'un indicateur (ex: "Population rurale")

plt.figure(figsize=(12, 6))
sns.lineplot(data=df1[df1['indicateur'] == 'Population rurale'], x='annee', y='valeur')
plt.title('Évolution Population rurale')
plt.xlabel('annee')
plt.ylabel('valeur')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Pour la dataframe df2

plt.figure(figsize=(12, 6))
sns.countplot(data=df2, x="sous_categorie", order=df2["sous_categorie"].value_counts().index)
plt.title("Distribution des sous-catégories")
plt.xlabel("Sous-catégorie")
plt.ylabel("Nombre d'observations")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# evolution de sous categorie dans le temps
plt.figure(figsize=(12, 6))
sns.lineplot(data=df2, x='annee', y='valeur', hue='sous_categorie')
plt.title('Évolution de la valeur des sous-catégories dans le temps')
plt.xlabel('Année')
plt.ylabel('Valeur')
plt.legend(title='Sous-catégorie', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid()
plt.show()
```
